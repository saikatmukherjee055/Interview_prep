```python
#Problem: Given a DataFrame containing information about employees, filter out the rows where 
#the employee’s age is greater than 30 and their salary is greater than $50,000.

import pandas as pd

# Sample DataFrame
data = {
    'Employee': ['John', 'Alice', 'Bob', 'Carol', 'Eve'],
    'Age': [25, 32, 28, 35, 45],
    'Salary': [48000, 54000, 52000, 58000, 65000]
}

df = pd.DataFrame(data)

# Filtering condition
filtered_df = df[(df['Age'] > 30) & (df['Salary'] > 50000)]

filtered_df
```

```python
#Problem: For a DataFrame containing sales data, group by the Region and calculate the total sales and average discount per region.

# Sample DataFrame
sales_data = {
    'Region': ['East', 'West', 'East', 'West', 'East', 'West'],
    'Sales': [234, 678, 123, 456, 789, 987],
    'Discount': [10, 20, 15, 10, 5, 25]
}

df = pd.DataFrame(sales_data)

# Grouping by 'Region' and aggregating
grouped_df = df.groupby('Region').agg({'Sales': 'sum', 'Discount': 'mean'})

grouped_df
```

```python
#Problem: Suppose you have a DataFrame containing information about products and their prices. 
#Create a new column that categorizes each product as either "Expensive" or "Cheap" based on a threshold price.

# Sample DataFrame
df = pd.DataFrame({
    'Product': ['A', 'B', 'C', 'D', 'E'],
    'Price': [200, 50, 150, 400, 30]
})

# Define the function to categorize products
def categorize(price):
    return 'Expensive' if price > 100 else 'Cheap'

# Apply the function to the 'Price' column
df['Category'] = df['Price'].apply(categorize)

df
```

```python
#Problem: Calculate the z-score of sales within each region for a DataFrame with regions and sales data.

# Sample DataFrame
df = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Sales': [234, 678, 123, 456, 789, 987]
})

# Z-score calculation function
def z_score(x):
    return (x - x.mean()) / x.std()

# Apply z-score transformation within each region
df['Z-Score'] = df.groupby('Region')['Sales'].transform(z_score)

df
```

```python
import pandas as pd

# Sample DataFrame
df_group = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South'],
    'Sales': [100, 200, 150, 250]
})

print(df_group)

# Apply z-score transformation within each region
z_scores = df_group.groupby('Region').apply(lambda x: (x['Sales'] - x['Sales'].mean()) / x['Sales'].std())

# Add the z-score column to the original DataFrame
df_group['Z-Score'] = z_scores.reset_index(drop=True)

print(df_group)
```

```python
#Problem: You have a DataFrame with phone numbers, and you want to identify which phone numbers match a specific pattern (e.g., XXX-XXX-XXXX).

# Sample DataFrame
df = pd.DataFrame({
    'PhoneNumber': ['123-456-7890', '987 654 3210', '1234567890', '(123)-456-7890', '123.456.7890']
})

print(df)


# Check if phone numbers match the pattern (XXX-XXX-XXXX)
df['Valid_Phone'] = df['PhoneNumber'].str.match(r'^\d{3}[-]\d{3}[-]\d{4}$')

print(df)

```

```python
#Problem: Extract all capitalized words from a column of text and store them in a new column.

# Sample DataFrame
df = pd.DataFrame({
    'Text': ['John Doe is a Data Scientist', 'Alice is an Engineer at Google', 'Bob works at IBM']
})

print(df)

# Extract all capitalized words using regex
df['Capitalized_Words'] = df['Text'].str.findall(r'\b[A-Z][a-z]*\b')

print(df)

```

```python
#Problem: Replace occurrences of multiple patterns in a column of text with a specified value.

# Sample DataFrame
df = pd.DataFrame({
    'Text': ['I like apples', 'I dislike bananas', 'I love cherries', 'I hate durians']
})

print(df)


# Replace 'dislike' and 'hate' with 'neutral'
df['Modified_Text'] = df['Text'].str.replace(r'dislike|hate', 'neutral', regex=True)

print(df)
```

```python
#Problem: Identify rows where the text starts and ends with a vowel.

# Sample DataFrame
df = pd.DataFrame({
    'Text': ['apple', 'banana', 'orange', 'umbrella', 'grape', 'kiwi']
})

print(df)

# Find strings that start and end with a vowel
df['Starts_Ends_Vowel'] = df['Text'].str.match(r'^[aeiouAEIOU].*[aeiouAEIOU]$')

print(df)


#Note:

'''
The . and * are distinct components of a regular expression, and they serve different purposes when combined. Let's break down the difference between using .* and using * alone.

What Does * Alone Mean?
The * symbol in a regular expression means "zero or more" of the preceding element. However, * must always follow a character, pattern, or character class to define what should be matched zero or more times.

a*: This means "zero or more a characters." It will match an empty string, "a", "aa", "aaa", and so on.
What Does . Mean?
The . (dot) in a regular expression matches any single character except a newline.

.: This matches exactly one character of any type (except a newline).
What Does .* Mean?
When you combine . and *, you get .*, which means "zero or more of any character."

.*: This will match any sequence of characters, including an empty sequence. It is used to match any string of characters between other specified patterns in a regex.
Why You Need Both . and *
If you used * by itself, without a preceding element, it wouldn't make sense in the context of a regular expression because * needs something to quantify. The * alone can't be used because it must be preceded by something that defines what should be repeated zero or more times.

Example with .*:
Consider the regex pattern ^[aeiouAEIOU].*[aeiouAEIOU]$:

[aeiouAEIOU]: Matches the first character as a vowel.
.*: Matches zero or more of any character(s) in between the first and last vowel.
[aeiouAEIOU]$: Matches the last character as a vowel.
This pattern will match strings like:

"apple"
"umbrella"
"AeIOU"
"a"
Example without .* (using * alone):
If you tried to use * alone, like this: ^[aeiouAEIOU]*[aeiouAEIOU]$, you would get a pattern that:

[aeiouAEIOU]*: Matches zero or more vowels at the start of the string.
[aeiouAEIOU]$: Matches a single vowel at the end of the string.
This pattern would match:

"a"
"aeiou"
"ai"
But it would not match:

"apple" (because ppl are not vowels and there's no .* to allow for these non-vowel characters in the middle).

Summary:
.*: This combination is necessary to match "zero or more of any character," allowing flexibility in matching any characters (or no characters) between specified patterns.
* alone: Cannot be used by itself; it needs a preceding character or pattern to define what should be matched zero or more times.
Thus, .* is essential when you need to allow for any characters to appear between the start and end patterns you're matching, like in the case of finding strings that start and end with a vowel.

'''
```

```python
import re

urls = "Visit https://www.example.com or http://another-example.org"
pattern = r'https?://(www\.)?([\w-]+)\.[\w.-]+'
result = re.findall(pattern, urls)
domains = [domain[1] for domain in result]
print(domains)  # Output: ['example', 'another-example']
```
